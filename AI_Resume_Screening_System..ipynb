```python
required_skills = [

    "Python",
    "Machine Learning",
    "SQL",
    "NLP",
    "Deep Learning",
    "Pandas",
    "Scikit-learn"
]

def extract_skills(resume):

    found_skills = []

    for skill in required_skills:

        if skill.lower() in resume.lower():

            found_skills.append(skill)

    return found_skills
```

```python
def match_skills(found_skills):

    matched = len(found_skills)

    total = len(required_skills)

    score = int((matched / total) * 100)

    return score
```

```python
explanation_template = PromptTemplate(

    input_variables=[

        "skills",
        "score"
    ],

    template="""

Candidate Skills:
{skills}

Fit Score:
{score}

Explain why this candidate received this score.
"""
)
```

```python
def evaluate_resume(resume):

    skills = extract_skills(resume)

    score = match_skills(skills)

    explanation = explanation_template.format(

        skills=skills,

        score=score
    )

    print("Extracted Skills:")
    print(skills)

    print("\nFit Score:")
    print(score)

    print("\nExplanation:")
    print(explanation)
```

```python
print("STRONG CANDIDATE\n")

evaluate_resume(strong_resume)
```

```python
print("AVERAGE CANDIDATE\n")

evaluate_resume(average_resume)
```

```python
print("WEAK CANDIDATE\n")

evaluate_resume(weak_resume)
```
